```python
#Exercise 1

#Define constants

h = 6.62607015e-34
c = 299792458
pi = 3.14159265359
hbar = h/(2 * pi)
e = 1.602176634e-19
me = 9.10938370e-31
ep0 = 8.85418781e-12

a0=4*pi*ep0*hbar**2/(me*e**2)
print('a0:' + str(a0))

#Calculate Energies

Eone=-e**2/(8*pi*ep0*a0)

Energies=[]

for n in range(1,6):
    En=Eone/n**2
    En_eV=En/e
    Energies.append(En_eV)
print('Energies:' + str(Energies))

Energies=set(Energies)
Energies=list(Energies)
Energies.sort(reverse=True)

Energy4=Energies[4]-Energies[2]
Energy3=Energies[3]-Energies[2]

#Calculate Wavelengths

Wavelength4=h*c/(Energy4*e)
Wavelength3=h*c/(Energy3*e)

print('Wavelength n=4 to n=2:' + str(Wavelength4) + 'm')

print('Wavelength n=3 to n=2:' + str(Wavelength3) + 'm')
```
